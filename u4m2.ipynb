```python
def wordFrequency(col, word):
    sdFreq = d2vDf[col].str.split(expand=True).stack().value_counts()
    sdFreq = pd.DataFrame(sdFreq).reset_index()
    sdFreq.columns = ['word','freq']
    sdFreq = sdFreq[sdFreq['word'].str.contains(word)]
    return sdFreq

print(wordFrequency('cleanDesc','advert'))
print(wordFrequency('cleanDesc','social' ))
print(wordFrequency('cleanDesc','search' ))
```
